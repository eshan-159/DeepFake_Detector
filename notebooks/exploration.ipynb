# Deepfake Detector Exploration
This scratch notebook demonstrates how to load a trained checkpoint, visualise a few predictions, and inspect Grad-CAM heatmaps.

```python
from pathlib import Path
from PIL import Image
import torch

from src.train.models.hybrid_cnn import HybridDeepfakeModel, ModelConfig

checkpoint = Path('../models/demo/demo.pt')
model = HybridDeepfakeModel(ModelConfig())
if checkpoint.exists():
    state = torch.load(checkpoint, map_location='cpu')
    model.load_state_dict(state if 'model' not in state else state['model'], strict=False)
    print('Loaded checkpoint', checkpoint)
else:
    print('Checkpoint missing — run training script first.')
```
